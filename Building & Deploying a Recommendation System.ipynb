```python
def convert_to_lowercase(text):
    return text.lower()
```

```python
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))
```

```python
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))
```

```python
def remove_stopwords(text):
    words = text.split()
    words = [word for word in words if word not in stop_words]
    return " ".join(words)
```

```python
df['clean_text'] = df['Overview'].apply(convert_to_lowercase)
```

```python
df['clean_text'] = df['clean_text'].apply(remove_punctuation)
```

```python
df['clean_text'] = df['clean_text'].apply(remove_stopwords)
```

## Task 3

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2)
)
```

```python
tfidf_matrix = tfidf.fit_transform(df['clean_text'])
```

```python
print("Shape of TF-IDF Matrix:", tfidf_matrix.shape)
```

```python
print("\nFirst 10 TF-IDF Features:")
print(tfidf.get_feature_names_out()[:10])
```

Observation
TF-IDF converted the 1000 cleaned movie overviews into numerical feature vectors.
Each row in the TF-IDF matrix represents a movie, while each column represents a unique word or phrase (feature).
max_features=5000 limits the vocabulary to the 5000 most important features, reducing dimensionality.
ngram_range=(1,2) includes both single words (unigrams) and two-word combinations (bigrams), capturing more context from the movie descriptions.
The resulting TF-IDF matrix will be used in the next task to calculate cosine similarity and generate movie recommendations.

## Task 4

Cosine similarity is used to measure how similar two movie descriptions are based on their TF-IDF vectors. It compares the angle between the vectors instead of their length, making it suitable for text data where documents may have different lengths. A higher cosine similarity score indicates that two movies have more similar content, so it is used to recommend movies with similar overviews.

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
cosine_sim = cosine_similarity(tfidf_matrix)
```

```python
print("Cosine Similarity Matrix Shape:", cosine_sim.shape)
```

```python
print(cosine_sim[0][:10])
```

Observation
The first similarity score is 1.0, which indicates that the movie is 100% similar to itself.
Most of the similarity scores are 0, indicating that those movies have no significant similarity in their descriptions.
A few movies have small similarity values such as 0.0399 and 0.0868, showing that they share some common words or themes with the selected movie.
The similarity score of 0.0868 is higher than 0.0399, meaning that movie is more closely related to the selected movie than the other.
Since the scores are relatively low, the selected movie has limited textual similarity with most other movies in the dataset based on their overviews.
These cosine similarity scores will be used to identify and recommend the most similar movies in the next task.

## Task 5

```python
# Recommendation function
def recommend(item_name, top_n=5):

    if item_name not in df["Series_Title"].values:
        return "Movie not found"

    index = df[df["Series_Title"] == item_name].index[0]

    similarity_scores = list(enumerate(cosine_sim[index]))

    similarity_scores = sorted(similarity_scores, key=lambda x: x[1], reverse=True)

    similarity_scores = similarity_scores[1:top_n+1]

    recommended_movies = []

    for movie in similarity_scores:
        recommended_movies.append(df.iloc[movie[0]]["Series_Title"])

    return recommended_movies
```

```python
recommend("Titanic")
```

```python
recommend("The Dark Knight")
```

```python
recommend("Avatar")
```

Observation
The recommendation function accepts a movie title as input and returns the top 5 most similar movies.
It uses the cosine similarity matrix to identify movies with the highest similarity scores.
The selected movie is excluded from the recommendations because it always has a similarity score of 1.0 with itself.
The recommended movies are based on similarities in their overview text, making this a content-based recommendation system.
If a movie title is not present in the dataset, the function returns "Movie not found!" to handle invalid input gracefully.
